```python
import sys

token_list = sys.stdin.read().split()
tokens = iter(token_list)

print("all tokens:", token_list)

test = int(next(tokens))

arr = [int(next(tokens)) for _ in range(test)]

row , col  = int(next(tokens)), int(next(tokens))
res = [[int(next(tokens)) for _ in range(col)] for _ in range(row)]

print("test:", test)
print("arr:", arr)
print("row:", row)
print("col:", col)
print("res:", res)
```

```python
import sys

class ListNode:
    def __init__(self , val= 0, next= None):
        self.val = val 
        self.next = next
        
def build_ll(arr):
    head = ListNode(arr[0])
    
    for i in range(1, len(arr)):
        curr = ListNode(arr[i])
        head.next = curr
        head = head.next
    return head
        
def print_linked_list(head):
    while head:
        print(head.val , end="")
        head = head.next 
    print()

def __name__ == "__main__":
    # 3 20 30 40 
    tokens = iter(sys.stdin.read().split())
    n = int(next(tokens))
    arr = [int(next(tokens)) for _ in range(n)]
    
    head = build_ll(arr)
    print_linked_list(head)
    
```
